```python
df=pd.read_csv('transit_columns3.csv')
df
```

```python
df_time = (df['max_time'] - df['median_time']) / (df['max_time'] - df['min_time'])
df_time
```

```python
df_distance = (df['최대 이동거리'] - df['이동거리 중간값']) / (df['최대 이동거리'] - df['최소 이동거리'])
df_distance
```

```python
df_change=(df['max_time'] - df['min_time']) / (df['max_time'] - (df['예상 배차시간'] + (df['환승횟수'] * df['예상 배차시간'])))
df_change
```

```python
df_transit=(df['최대 교통수단 수']/df['num_trainsit_methods'])
df_transit
```

```python
# 가중치 설정
w_distance = 0.4        #거리 접근성 가중치
w_time = 0.3    # 시간 접근성 가중치
w_환승 = 0.2        # 환승 접근성 가중치
w_교통수단 = 0.1    # 교통수단 접근성 가중치
```

```python
df_composite = (
    w_time * df_time +
    w_distance * df_distance +
    w_환승 * df_change +
    w_교통수단 * df_transit
)
```
